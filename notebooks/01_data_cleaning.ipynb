```python
import pandas as pd

soil_df = pd.read_csv("../data/raw/state_soil_data.csv")
soil_df.head()
```

```python
soil_df["state"] = soil_df["state"].str.strip().str.lower()
```

```python
soil_df = soil_df.drop_duplicates()
```

```python
soil_df.to_csv(
    "../data/processed/state_soil_data_cleaned.csv",
    index=False
)
```

```python
weather_df = pd.read_csv("../data/raw/state_weather_data_1997_2020.csv")
weather_df.head()
```

```python
weather_df.shape
weather_df.info()
weather_df.isnull().sum()
```

```python
weather_df.columns = (
    weather_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

weather_df.columns
```

```python
weather_df["state"] = weather_df["state"].str.strip().str.lower()
```

```python
weather_df["year"].describe()
```

```python
weather_df.select_dtypes(include=["float64", "int64"]).columns
```

```python
for col in ["temperature", "rainfall", "humidity"]:
    if col in weather_df.columns:
        weather_df[col] = weather_df[col].fillna(weather_df[col].mean())
```

```python
weather_df = weather_df.drop_duplicates()
```

```python
weather_agg = (
    weather_df
    .groupby(["state", "year"], as_index=False)
    .mean(numeric_only=True)
)
```

```python
weather_agg.to_csv(
    "../data/processed/state_weather_data_cleaned.csv",
    index=False
)
```

```python
yield_df = pd.read_csv("../data/raw/crop_yield.csv")
yield_df.head()
```

```python
yield_df.columns = (
    yield_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

yield_df.columns
```

```python
yield_df["state"] = yield_df["state"].str.strip().str.lower()
yield_df["crop"] = yield_df["crop"].str.strip().str.lower()
yield_df["season"] = yield_df["season"].str.strip().str.lower()
```

```python
yield_df.describe()
```

```python
yield_df = yield_df[yield_df["yield"] > 0]
yield_df = yield_df[yield_df["area"] > 0]
```

```python
yield_df = yield_df[yield_df["production"] >= 0]
```

```python
yield_df = yield_df.drop_duplicates()
```

```python
yield_df["yield"] = yield_df["yield"].round(2)
yield_df["area"] = yield_df["area"].round(2)
```

```python
yield_df.shape
yield_df.info()
```

```python
yield_df.to_csv(
    "../data/processed/crop_yield_cleaned.csv",
    index=False
)
```

```python
import pandas as pd

crop_df = pd.read_csv("../data/processed/crop_recommendation_cleaned.csv")
soil_df = pd.read_csv("../data/processed/state_soil_data_cleaned.csv")
weather_df = pd.read_csv("../data/processed/state_weather_data_cleaned.csv")
yield_df = pd.read_csv("../data/processed/crop_yield_cleaned.csv")
```

```python
crop_df.shape, soil_df.shape, weather_df.shape, yield_df.shape
```

```python
for df in [soil_df, weather_df, yield_df]:
    df["state"] = df["state"].str.lower().str.strip()
```

```python
soil_weather_df = pd.merge(
    soil_df,
    weather_df,
    on=["state"],
    how="inner"
)
```

```python
soil_weather_df.shape
soil_weather_df.head()
```

```python
final_df = pd.merge(
    yield_df,
    soil_weather_df,
    on=["state", "year"],
    how="inner"
)
```

```python
final_df.shape
final_df.head()
```

```python
final_df.to_csv(
    "../data/processed/final_combined_dataset.csv",
    index=False
)
```
