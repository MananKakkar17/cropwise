```python
import pandas as pd

df = pd.read_csv("../data/processed/final_combined_dataset.csv")
df.head()
```

```python
df_model = df.drop(
    columns=["yield", "production", "area", "fertilizer", "pesticide"]
)
df_model.head()
```

```python
num_features = [
    "n", "p", "k", "soil_ph",
    "avg_temp_c",
    "total_rainfall_mm",
    "avg_humidity_percent"
]
```

```python
cat_features = ["season"]
```

```python
df_model = pd.get_dummies(
    df_model,
    columns=cat_features,
    drop_first=True
)
```

```python
df_model.columns
```

```python
df_model.columns = df_model.columns.str.replace(" ", "_")
df_model.columns
```

```python
X = df_model.drop(columns=["crop"])
y = df_model["crop"]
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)
```

```python
label_mapping = dict(zip(label_encoder.classes_, label_encoder.transform(label_encoder.classes_)))
label_mapping
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
X_scaled = X.copy()

X_scaled[num_features] = scaler.fit_transform(X[num_features])
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y_encoded,
    test_size=0.2,
    random_state=42,
    stratify=y_encoded
)
```
